```python
def find_zero_derivative(start_x, h, tolerance = 1e-6, max_steps = 100000000):

    x = start_x
    x_list = []
    for _ in range(max_steps):
        derivative = (f(x + h) - f(x)) / h
        x_list.append(x)

        if abs(derivative) < tolerance:
            print(x)
            return x_list
        x = x - (h * derivative)

    print("reached max_steps. x = ", x)
    
    return x_list
```

```python
x_dash = find_zero_derivative(10, 0.001)
y_values = [f(x) for x in x_dash]

plt.plot(x_dash, y_values)
```
